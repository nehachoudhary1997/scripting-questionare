```python
#. Filter Requests with 5xx Status Codes
 
import re
status_pattern=re.compile(r'\b5\d{2}\b')

with open("access.log" ,encoding="utf-8") as file:
    for line in file:
        if status_pattern.search(line):
            print(line.rstrip('\n'))

```

```python
#adding Save all 5xx log lines into a new file named errors.log instead of printing them.

#. Filter Requests with 5xx Status Codes
 
import re
status_pattern=re.compile(r'\b5\d{2}\b')
with open("access.log" ,encoding="utf-8") as file:
    for line in file:
        if status_pattern.search(line):
           with open("error.log","w") as f:
            f.write(line)
```

Why shouldn't we open the file inside the loop?

A good answer is:

Opening and closing a file is an expensive I/O operation. If I open it inside the loop, it happens once for every matching line. By opening the file once before the loop, I avoid repeated I/O overhead and, in "w" mode, I also avoid accidentally overwriting previously written data.

```python
import re
status_pattern=re.compile(r'"\s+(5\d{2})')
with open("app.log") as infile , open("error.log",'w') as outfile:
    for line in infile:
        if status_pattern.search(line):
            outfile.write(line)
```
