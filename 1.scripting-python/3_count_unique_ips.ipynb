```python
#Q3. Count Number of Unique IPs Making Requests
#Task: Count how many unique IP addresses accessed the server.
import re

ips=set()
ip_pattern = re.compile(r'\b(?:\d{1,3}\.){3}\d{1,3}\b')
with open("access.log") as file:
    for line in file:
        for i in ip_pattern.findall(line):
            ips.add(i)
print(len(ips))
```

```python
#Q3. Count Number of Unique IPs Making Requests
#Task: Count how many unique IP addresses accessed the server.
import re

ips=set()
ip_pattern = re.compile(r'\b(?:\d{1,3}\.){3}\d{1,3}\b')
with open("app.log") as file:
    for line in file:
        ips.update(ip_pattern.findall(line)) #add multi ele
        
print(len(ips))
```

For your log-parsing questions:

✅ One IP per line (standard Nginx/Apache logs): search() + add()
✅ Possibly multiple IPs per line: findall() + update()
