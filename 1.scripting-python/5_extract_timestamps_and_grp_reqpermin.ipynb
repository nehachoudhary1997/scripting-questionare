----
Q5. Extract Timestamp and Group Requests Per Minute
Task: Extract timestamps and count number of requests per minute.


203.175.100.25 - - [19/May/2026 10:15:30] "GET /trending HTTP/1.1" 500 3456
timestamp- 19/May/2026 
group requests time by  10:15:30 (hr:min:sec)
----

```python
import re
pattern=re.compile(r'\[(\S+)\s+(\d{1,2}\:\d{1,2}):]')
with open("Q5.log") as file:
    for line in file:
        match=pattern.search(line)
        if match:
            timestamp=match.group(1)
            grp_request_time=match.group(2)
            print(timestamp,grp_request_time)
```

----
Why is this useful?

Suppose your server normally receives:

10:20 -> 35 requests
10:21 -> 38 requests
10:22 -> 250 requests

Immediately you know something unusual happened at 10:22.

Possible reasons:

A traffic spike
A DDoS attack
A marketing campaign sending many users
A monitoring script hitting the server repeatedly

This kind of grouping is very common in monitoring tools like Grafana, Prometheus, Splunk, and the ELK stack.
