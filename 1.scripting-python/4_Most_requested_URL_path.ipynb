Q4. Identify the Most Frequently Requested URL Path
Log Line:
 203.0.113.42 - - [24/May/2025:10:22:15 +0000] "GET /api/user/profile HTTP/1.1" 200 350
 Task: Extract the path (like /api/user/profile) and count the most accessed one.


```python
import re
from collections import Counter
url_pattern=re.compile(r'"[A-Z]+\s+(\S+)\s+HTTP')
url=""
count=Counter()
with open("app.log") as file:
    for line in file:
        match=url_pattern.search(line)
        if match:
            url=match.group(1)
            count[url]+=1
            
for path,c in count.most_common(1):
    print(path,c)
```

```python
#all urls sorted
for path, c in count.most_common():
    print(path, c)
```

```python
#Ignore query parameters. eg-GET /search?q=python&page=2 HTTP/1.1 
import re
from collections import Counter
url_pattern=re.compile(r'"[A-Z]+\s+(\S+)\s+HTTP')
url=""
count=Counter()
with open("app.log") as file:
    for line in file:
        match=url_pattern.search(line)
        if match:
            url = match.group(1).split("?")[0]
            count[url] += 1

for p,c in count.most_common():
    print(p,c)
```
